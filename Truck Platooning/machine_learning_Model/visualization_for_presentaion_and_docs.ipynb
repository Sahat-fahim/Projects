```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("Truck_data_Labeled_random_leaders.csv")
df.head()
```

```python
X = df[['Leadership_score']]
X.head()
```

```python
Y = df.Leader
Y.head()
```

```python
plt.scatter(X[:500], Y[:500], color='red')  # Select the first 500 data points
plt.xlabel('Leadership score')
plt.ylabel('Leader')
plt.title('Data Points Plot')
plt.grid(True)
plt.legend()
plt.show()
```
